```python


import uvicorn
from fastapi import FastAPI
```

## Sobre FASTAPI

* FastAPI é uma forma rápida de criar uma conexão de API, onde eu posso criar um servidor para colocar minhas funções. 
* As funções devem ser informadas com GET ou POST.
* GET - Pega as informações do server
* POST - cliente manda informações, e depois recebe a resposta.

```python
# Exemplo de como criar uma API simples com FastAPI
app = fastapi.FastAPI()  # Instanciando a aplicação FastAPI

@app.get("/soma")  # Definine uma rota GET para a função soma
def soma(a: int, b: int):  # Função que recebe dois parâmetros inteiros
    return {"resultado": a + b}  # Retorna a soma dos dois números em formato JSON
```

## Sobre FastAPI

* Aqui quero criar um servidor local, direto no notebook.
Após rodar a célular podemos acessar o servidor de duas formas:
1. http://127.0.0.1:8000/soma?a=3&b=5  - Neste caso estou colocando o servidor, chamando a função e dando os valores.
2. http://127.0.0.1:8000/docs - Neste caso consigo "entrar no site" e testar direto no Browser o resultado.

* Após rodar o servidor, as células ficaram bloqueadas porque a célular do servidor rodará até ser parada.

```python
config = uvicorn.Config(app, host="0.0.0.0", port=8000)  # Cria o objeto de configuração do servidor Uvicorn
server = uvicorn.Server(config)     # Cria o objeto do servidor Uvicorn com a configuração especificada

await server.serve()   #Roda o servidor Uvicorn de forma assíncrona, permitindo que ele aceite conexões e processe requisições.
```

## Abaixo vou rodar em thread
* Isso não trava a célula

```python
import threading

def run():
    uvicorn.run(app, host="0.0.0.0", port=8000)

threading.Thread(target=run, daemon=True).start()
```

# Vamos criar o cliente
* Esta sessão de cliente é a forma como iremos consumir a API do servidor
* Vamos usar o requests

```python
import requests
```

```python
base_url = "http://localhost:8000"  # URL base da API
endpoint = "/soma"  # Endpoint da função soma
params = {"a": 5, "b": 3}  # Parâmetros

url = f"{base_url}{endpoint}"  # Monta a URL completa
response = requests.get(url, params=params)  # Faz a requisição GET para a API
```

```python
response.json() # Exibe a resposta da API em formato JSON
```
